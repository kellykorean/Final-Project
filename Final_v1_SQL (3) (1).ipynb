## SQL Analysis

This section contains the required SQL queries for the final project. Each query includes comments explaining what it does, how it works, and why it matters to the analysis.

```python
import pandas as pd
import sqlite3

conn = sqlite3.connect("Final_v1.db")
```

```python
merged_df.to_sql(
    "country_metrics",
    conn,
    if_exists="replace",
    index=False
)
```

```python
pd.read_sql("""
-- Query 1: Preview country metrics table
-- Purpose: Confirm that the final dataset loaded into SQLite correctly.
-- Method: Select the first 10 rows from the country_metrics table.
-- Why: This gives a quick check that the SQL table contains the expected country-level data.

SELECT *
FROM country_metrics
LIMIT 10;
""", conn)
```

```python
pd.read_sql("""
-- Query 2: Check missing values in key fields
-- Purpose: Identify missing values in GDP, HDI, latitude, and climate zone.
-- Method: Use CASE WHEN statements to count missing values by column.
-- Why: Missing values can affect EDA, modeling, and conclusions.

SELECT
    SUM(CASE WHEN gdp_per_capita IS NULL THEN 1 ELSE 0 END) AS missing_gdp,
    SUM(CASE WHEN hdi_value IS NULL THEN 1 ELSE 0 END) AS missing_hdi,
    SUM(CASE WHEN latitude IS NULL THEN 1 ELSE 0 END) AS missing_latitude,
    SUM(CASE WHEN climate_zone IS NULL THEN 1 ELSE 0 END) AS missing_climate_zone
FROM country_metrics;
""", conn)
```

```python
pd.read_sql("""
-- Query 3: Check unique regions and climate zones
-- Purpose: Review the categorical variables used in the analysis.
-- Method: Count distinct regions and climate zones.
-- Why: This confirms that the dataset has meaningful geographic groupings for EDA.

SELECT
    COUNT(DISTINCT region) AS unique_regions,
    COUNT(DISTINCT climate_zone) AS unique_climate_zones
FROM country_metrics;
""", conn)
```

```python
pd.read_sql("""
-- Query 4: GDP and HDI summary by region
-- Purpose: Compare economic and development outcomes across regions.
-- Method: Group by region and calculate country count, average GDP, and average HDI.
-- Why: Regional comparisons help identify broad geographic patterns in development.

SELECT
    region,
    COUNT(*) AS country_count,
    AVG(gdp_per_capita) AS avg_gdp_per_capita,
    AVG(hdi_value) AS avg_hdi
FROM country_metrics
GROUP BY region
ORDER BY avg_gdp_per_capita DESC;
""", conn)
```

```python
pd.read_sql("""
-- Query 5: GDP and HDI summary by climate zone
-- Purpose: Compare GDP and HDI across climate zones.
-- Method: Group by climate_zone and calculate average GDP and HDI.
-- Why: This directly connects to the project question about geography and development outcomes.

SELECT
    climate_zone,
    COUNT(*) AS country_count,
    AVG(gdp_per_capita) AS avg_gdp_per_capita,
    AVG(hdi_value) AS avg_hdi,
    AVG(abs_latitude) AS avg_abs_latitude
FROM country_metrics
GROUP BY climate_zone
ORDER BY avg_gdp_per_capita DESC;
""", conn)
```

```python
pd.read_sql("""
-- Query 6: Join geography and development views
-- Purpose: Demonstrate how country geography connects to development outcomes.
-- Method: Create geography and development subqueries, then join them on iso_code.
-- Why: This separates geographic variables from development variables and recombines them for analysis.

SELECT
    g.country,
    g.region,
    g.climate_zone,
    g.abs_latitude,
    d.gdp_per_capita,
    d.hdi_value
FROM
    (
        SELECT iso_code, country, region, climate_zone, abs_latitude
        FROM country_metrics
    ) AS g
JOIN
    (
        SELECT iso_code, gdp_per_capita, hdi_value
        FROM country_metrics
    ) AS d
ON g.iso_code = d.iso_code
ORDER BY d.gdp_per_capita DESC
LIMIT 20;
""", conn)
```

```python
pd.read_sql("""
-- Query 7: Rank countries by GDP within region
-- Purpose: Identify the highest GDP countries within each region.
-- Method: Join geography and economic subqueries, then use RANK() over each region.
-- Why: Ranking within region gives a more meaningful peer comparison than global ranking alone.

SELECT
    g.country,
    g.region,
    e.gdp_per_capita,
    RANK() OVER (
        PARTITION BY g.region
        ORDER BY e.gdp_per_capita DESC
    ) AS regional_gdp_rank
FROM
    (
        SELECT iso_code, country, region
        FROM country_metrics
    ) AS g
JOIN
    (
        SELECT iso_code, gdp_per_capita
        FROM country_metrics
    ) AS e
ON g.iso_code = e.iso_code
ORDER BY g.region, regional_gdp_rank;
""", conn)
```

```python
pd.read_sql("""
-- Query 8: Compare each country's GDP to its climate-zone average
-- Purpose: Show whether countries are above or below the average GDP of their climate zone.
-- Method: Use AVG() OVER PARTITION BY climate_zone to calculate climate-zone averages.
-- Why: This helps identify outliers and exceptions within each geographic category.

SELECT
    country,
    climate_zone,
    gdp_per_capita,
    AVG(gdp_per_capita) OVER (
        PARTITION BY climate_zone
    ) AS climate_zone_avg_gdp,
    gdp_per_capita - AVG(gdp_per_capita) OVER (
        PARTITION BY climate_zone
    ) AS difference_from_climate_zone_avg
FROM country_metrics
ORDER BY difference_from_climate_zone_avg DESC;
""", conn)
```

```python
pd.read_sql("""
-- Query 9: Countries above sample average GDP per capita
-- Purpose: Identify countries with GDP per capita above the sample average.
-- Method: Use a subquery to calculate average GDP per capita, then filter countries above it.
-- Why: This helps distinguish higher-income countries from the overall country sample.

SELECT
    country,
    region,
    climate_zone,
    gdp_per_capita,
    hdi_value
FROM country_metrics
WHERE gdp_per_capita > (
    SELECT AVG(gdp_per_capita)
    FROM country_metrics
)
ORDER BY gdp_per_capita DESC;
""", conn)
```

```python
pd.read_sql("""
-- Query 10: Countries above sample average HDI
-- Purpose: Identify countries with Human Development Index values above the sample average.
-- Method: Use a subquery to calculate average HDI, then filter countries above it.
-- Why: This helps compare economic prosperity with broader human development outcomes.

SELECT
    country,
    region,
    climate_zone,
    gdp_per_capita,
    hdi_value
FROM country_metrics
WHERE hdi_value > (
    SELECT AVG(hdi_value)
    FROM country_metrics
)
ORDER BY hdi_value DESC;
""", conn)
```
